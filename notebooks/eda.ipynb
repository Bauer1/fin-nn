# fin-nn: Exploratory Data Analysis (NVDA)

Overview of raw NVDA data (2018–2026), verification of the 10-for-1 stock split (June 2024), distributions of the 6 technical features, and class balance across `train`, `val`, and `test`.

```python
import matplotlib.pyplot as plt
import pandas as pd

from src.data import get_data
from src.features import FEATURE_COLS, TARGET_COL, prepare_data

raw_df = get_data()
print(f"Raw shape: {raw_df.shape}, period: {raw_df.index[0].date()} -> {raw_df.index[-1].date()}")
raw_df.tail()
```

```python
fig, ax = plt.subplots(figsize=(12, 4))
raw_df["Close"].plot(ax=ax, title="NVDA Adjusted Close Price (2018-2026)", lw=1.5)
ax.set_ylabel("Price ($)")
ax.grid(alpha=0.3)
plt.show()
```

```python
split_data = prepare_data()
for name, part in [("Train", split_data.train_df), ("Val", split_data.val_df), ("Test", split_data.test_df)]:
    print(f"{name:5s}: {len(part):4d} rows | Up ratio: {part[TARGET_COL].mean():.2%}")

split_data.train_df[FEATURE_COLS].describe().T
```
